```python
import math

import numpy as np
from math import sqrt
```

```python
semi_minor = 5003
semi_major = 12003

Q = np.diag([
    1/semi_minor**2,
    1/semi_major**2,
    1/semi_minor**2,
    -1
])
Q
```

```python
E_plane = np.array([
    [0, 1, 0],
    [0, 0, semi_major-1],
    [1, 0, 0],
    [0, 0, 1]
])

E_plane
```

```python
C = np.linalg.matrix_transpose(E_plane) @ Q @ E_plane
C
```

```python
np.linalg.det(C)
```

```python
np.linalg.det(C[:2,:2])
```

```python
scale_sphere_down = np.diag([1/semi_minor, 1/semi_major, 1/semi_minor, 1])

E_plane_scaled = scale_sphere_down @ E_plane
E_plane_scaled
```

```python
plane_norm_scaled = np.linalg.cross(E_plane_scaled[:3, 0], E_plane_scaled[:3, 1])
plane_norm_scaled
```

```python
plane_d = plane_norm_scaled @ E_plane_scaled[:3, 2]
plane_d
```

```python
plane_distance = abs(plane_d) / np.linalg.norm(plane_norm_scaled)
plane_distance
```

```python
plane_d ** 2
```

```python
plane_d ** 2 < plane_norm_scaled @  plane_norm_scaled
```

```python
def greater_than_or_close(x: float, y: float) -> bool:
    return x > y or math.isclose(x, y)
```

```python
not greater_than_or_close(abs(plane_d), np.linalg.norm(plane_norm_scaled))
```
